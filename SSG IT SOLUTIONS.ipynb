```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
# Load the South African education and unemployment datasets

education = pd.read_csv(
    "../datasets/education/API_SE.TER.ENRR_DS2_en_csv_v2_34068.csv",
    skiprows=4
)

unemployment = pd.read_csv(
    "../datasets/unemployment/API_SL.UEM.TOTL.ZS_DS2_en_csv_v2_33398.csv",
    skiprows=4
)

print("Education dataset loaded successfully!")
print("Unemployment dataset loaded successfully!")
```

```python
print("Education Dataset Shape:", education.shape)
print("Unemployment Dataset Shape:", unemployment.shape)

print("\nEducation Dataset Columns:")
print(education.columns.tolist())

print("\nUnemployment Dataset Columns:")
print(unemployment.columns.tolist())
```

```python
# Remove unnecessary unnamed columns

education = education.loc[
    :, ~education.columns.str.contains("^Unnamed")
]

unemployment = unemployment.loc[
    :, ~unemployment.columns.str.contains("^Unnamed")
]

print("Unnecessary columns removed successfully!")

print("\nEducation shape:", education.shape)
print("Unemployment shape:", unemployment.shape)
```

```python
# Extract South Africa from both datasets

education_sa = education[
    education["Country Name"] == "South Africa"
].reset_index(drop=True)

unemployment_sa = unemployment[
    unemployment["Country Name"] == "South Africa"
].reset_index(drop=True)

print("South Africa education data:")
display(education_sa)

print("\nSouth Africa unemployment data:")
display(unemployment_sa)
```

```python
# Select only the yearly values

education_years = education_sa.iloc[:, 4:]
unemployment_years = unemployment_sa.iloc[:, 4:]

print("Education yearly data:")
display(education_years)

print("\nUnemployment yearly data:")
display(unemployment_years)
```

```python
# Convert yearly values to numeric

education_years = education_years.apply(
    pd.to_numeric,
    errors="coerce"
)

unemployment_years = unemployment_years.apply(
    pd.to_numeric,
    errors="coerce"
)

print("Education yearly data converted to numeric.")
print("Unemployment yearly data converted to numeric.")

print("\nEducation data types:")
print(education_years.dtypes)

print("\nUnemployment data types:")
print(unemployment_years.dtypes)
```

```python
# Check for missing values

print("Missing values in Education:")
print(education_years.isnull().sum())

print("\nMissing values in Unemployment:")
print(unemployment_years.isnull().sum())
```

```python
# Transpose the yearly data

education_data = education_years.T
education_data.columns = ["Education"]

unemployment_data = unemployment_years.T
unemployment_data.columns = ["Unemployment"]

print("Datasets transposed successfully!")

print("\nEducation data:")
display(education_data.head())

print("\nUnemployment data:")
display(unemployment_data.head())
```

```python
# Convert the index into a Year column

education_data.reset_index(inplace=True)
education_data.rename(
    columns={"index": "Year"},
    inplace=True
)

unemployment_data.reset_index(inplace=True)
unemployment_data.rename(
    columns={"index": "Year"},
    inplace=True
)

print("Year column created successfully!")

print("\nEducation data:")
display(education_data.head())

print("\nUnemployment data:")
display(unemployment_data.head())
```

```python
# Convert Year to numeric

education_data["Year"] = pd.to_numeric(
    education_data["Year"],
    errors="coerce"
)

unemployment_data["Year"] = pd.to_numeric(
    unemployment_data["Year"],
    errors="coerce"
)

# Remove rows where Year could not be converted

education_data = education_data.dropna(subset=["Year"])
unemployment_data = unemployment_data.dropna(subset=["Year"])

# Convert Year to integer

education_data["Year"] = education_data["Year"].astype(int)
unemployment_data["Year"] = unemployment_data["Year"].astype(int)

print("Year converted successfully!")

display(education_data.head())
```

```python
# Remove missing observations

education_data = education_data.dropna(
    subset=["Education"]
)

unemployment_data = unemployment_data.dropna(
    subset=["Unemployment"]
)

print("Missing observations removed.")

print("\nEducation data shape:", education_data.shape)
print("Unemployment data shape:", unemployment_data.shape)
```

```python
# Combine Education and Unemployment using Year

combined_data = pd.merge(
    education_data,
    unemployment_data,
    on="Year",
    how="inner"
)

# Sort by Year

combined_data = combined_data.sort_values(
    "Year"
).reset_index(drop=True)

print("Datasets combined successfully!")

display(combined_data)
```

```python
# Final combined dataset check

print("=" * 60)
print("FINAL COMBINED DATASET")
print("=" * 60)

print("\nShape:")
print(combined_data.shape)

print("\nData types:")
print(combined_data.dtypes)

print("\nMissing values:")
print(combined_data.isnull().sum())

print("\nFinal dataset:")
display(combined_data)
```

```python
# Check missing values before cleaning

missing_values = pd.DataFrame({
    "Education": education_data["Education"].isna().sum(),
    "Unemployment": unemployment_data["Unemployment"].isna().sum()
}, index=["Missing Values"])

display(missing_values)
```

```python
# Identify years with missing education values

education_missing = education_data[
    education_data["Education"].isna()
]

print("Years with missing Education values:")
display(education_missing)
```

```python
# Identify years with missing unemployment values

unemployment_missing = unemployment_data[
    unemployment_data["Unemployment"].isna()
]

print("Years with missing Unemployment values:")
display(unemployment_missing)
```

```python
# Record number of observations before removing missing values

education_before = len(education_data)
unemployment_before = len(unemployment_data)

# Remove missing observations

education_clean = education_data.dropna(
    subset=["Education"]
).copy()

unemployment_clean = unemployment_data.dropna(
    subset=["Unemployment"]
).copy()

print("Education observations before cleaning:", education_before)
print("Education observations after cleaning:", len(education_clean))

print("\nUnemployment observations before cleaning:", unemployment_before)
print("Unemployment observations after cleaning:", len(unemployment_clean))
```

```python
# Combine the cleaned datasets

combined_data = pd.merge(
    education_clean,
    unemployment_clean,
    on="Year",
    how="inner"
)

combined_data = combined_data.sort_values(
    "Year"
).reset_index(drop=True)

print("Final combined dataset created.")

print("\nShape:", combined_data.shape)

display(combined_data)
```

```python
# Final missing-value check

print("Missing values in final dataset:")

display(combined_data.isnull().sum())
```

```python
# Descriptive statistics using Pandas

descriptive_stats = combined_data[
    ["Education", "Unemployment"]
].describe()

display(descriptive_stats)
```

```python
# Save the final cleaned and combined dataset

combined_data.to_csv(
    "../datasets/combined_data.csv",
    index=False
)

print("Final cleaned dataset saved successfully.")
```

```python
# Create NumPy arrays from the combined dataset

education_array = combined_data["Education"].to_numpy()
unemployment_array = combined_data["Unemployment"].to_numpy()
year_array = combined_data["Year"].to_numpy()

print("Education NumPy array:")
print(education_array)

print("\nUnemployment NumPy array:")
print(unemployment_array)

print("\nYear NumPy array:")
print(year_array)
```

```python
# Inspect NumPy array dimensions and shapes

print("Education array shape:", education_array.shape)
print("Education array dimensions:", education_array.ndim)

print("\nUnemployment array shape:", unemployment_array.shape)
print("Unemployment array dimensions:", unemployment_array.ndim)

print("\nYear array shape:", year_array.shape)
print("Year array dimensions:", year_array.ndim)
```

```python
# Create a two-dimensional NumPy array

analysis_array = np.column_stack(
    (education_array, unemployment_array)
)

print("2D analysis array:")
print(analysis_array)

print("\nShape:", analysis_array.shape)
print("Dimensions:", analysis_array.ndim)
```

```python
# Reshape the analysis array

reshaped_array = analysis_array.reshape(
    -1,
    2
)

print("Original shape:", analysis_array.shape)
print("Reshaped shape:", reshaped_array.shape)

print("\nReshaped array:")
print(reshaped_array)
```

```python
# Calculate descriptive statistics using NumPy

education_mean_np = np.mean(education_array)
education_median_np = np.median(education_array)
education_std_np = np.std(education_array)
education_min_np = np.min(education_array)
education_max_np = np.max(education_array)

unemployment_mean_np = np.mean(unemployment_array)
unemployment_median_np = np.median(unemployment_array)
unemployment_std_np = np.std(unemployment_array)
unemployment_min_np = np.min(unemployment_array)
unemployment_max_np = np.max(unemployment_array)

print("EDUCATION — NumPy")
print("-" * 45)
print(f"Mean: {education_mean_np:.2f}")
print(f"Median: {education_median_np:.2f}")
print(f"Standard deviation: {education_std_np:.2f}")
print(f"Minimum: {education_min_np:.2f}")
print(f"Maximum: {education_max_np:.2f}")

print("\nUNEMPLOYMENT — NumPy")
print("-" * 45)
print(f"Mean: {unemployment_mean_np:.2f}")
print(f"Median: {unemployment_median_np:.2f}")
print(f"Standard deviation: {unemployment_std_np:.2f}")
print(f"Minimum: {unemployment_min_np:.2f}")
print(f"Maximum: {unemployment_max_np:.2f}")
```

```python
# Calculate the range of each indicator

education_range = np.ptp(education_array)
unemployment_range = np.ptp(unemployment_array)

print(f"Education range: {education_range:.2f}")
print(f"Unemployment range: {unemployment_range:.2f}")
```

```python
# Calculate year-to-year changes

education_changes = np.diff(education_array)
unemployment_changes = np.diff(unemployment_array)

print("Year-to-year Education changes:")
print(education_changes)

print("\nYear-to-year Unemployment changes:")
print(unemployment_changes)
```

```python
# Calculate year-to-year percentage changes

education_percentage_change = (
    np.diff(education_array) / education_array[:-1]
) * 100

unemployment_percentage_change = (
    np.diff(unemployment_array) / unemployment_array[:-1]
) * 100

print("Education percentage changes:")
print(np.round(education_percentage_change, 2))

print("\nUnemployment percentage changes:")
print(np.round(unemployment_percentage_change, 2))
```

```python
# Find the largest year-to-year changes

education_largest_increase_index = np.argmax(
    education_changes
)

education_largest_decrease_index = np.argmin(
    education_changes
)

unemployment_largest_increase_index = np.argmax(
    unemployment_changes
)

unemployment_largest_decrease_index = np.argmin(
    unemployment_changes
)

print("Education largest increase:")
print(
    f"{year_array[education_largest_increase_index]} "
    f"to "
    f"{year_array[education_largest_increase_index + 1]}: "
    f"{education_changes[education_largest_increase_index]:.2f}"
)

print("\nEducation largest decrease:")
print(
    f"{year_array[education_largest_decrease_index]} "
    f"to "
    f"{year_array[education_largest_decrease_index + 1]}: "
    f"{education_changes[education_largest_decrease_index]:.2f}"
)

print("\nUnemployment largest increase:")
print(
    f"{year_array[unemployment_largest_increase_index]} "
    f"to "
    f"{year_array[unemployment_largest_increase_index + 1]}: "
    f"{unemployment_changes[unemployment_largest_increase_index]:.2f}"
)

print("\nUnemployment largest decrease:")
print(
    f"{year_array[unemployment_largest_decrease_index]} "
    f"to "
    f"{year_array[unemployment_largest_decrease_index + 1]}: "
    f"{unemployment_changes[unemployment_largest_decrease_index]:.2f}"
)
```

```python
# Calculate the correlation between education and unemployment

correlation_matrix = np.corrcoef(
    education_array,
    unemployment_array
)

print("NumPy correlation matrix:")
print(correlation_matrix)

correlation = correlation_matrix[0, 1]

print(
    f"\nCorrelation coefficient: {correlation:.4f}"
)
```

```python
# Create a NumPy numerical analysis summary

numpy_summary = pd.DataFrame({
    "Education": [
        education_mean_np,
        education_median_np,
        education_std_np,
        education_min_np,
        education_max_np,
        education_range
    ],
    "Unemployment": [
        unemployment_mean_np,
        unemployment_median_np,
        unemployment_std_np,
        unemployment_min_np,
        unemployment_max_np,
        unemployment_range
    ]
}, index=[
    "Mean",
    "Median",
    "Standard Deviation",
    "Minimum",
    "Maximum",
    "Range"
])

display(numpy_summary.round(2))
```

```python
# Import SQLite

import sqlite3
from pathlib import Path

print("SQLite imported successfully.")
```

```python
# Create the SQLite database

database_path = Path("../database/ndta631_analysis.db")

connection = sqlite3.connect(database_path)

cursor = connection.cursor()

print("Database connection established successfully.")
print("Database:", database_path)
```

```python
# Create the combined_data table

cursor.execute("""
CREATE TABLE IF NOT EXISTS combined_data (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    year INTEGER NOT NULL UNIQUE,
    education REAL,
    unemployment REAL
)
""")

connection.commit()

print("combined_data table created successfully.")
```

```python
# Insert the cleaned combined dataset into SQLite

insert_query = """
INSERT OR IGNORE INTO combined_data
(year, education, unemployment)
VALUES (?, ?, ?)
"""

for _, row in combined_data.iterrows():
    cursor.execute(
        insert_query,
        (
            int(row["Year"]),
            float(row["Education"]),
            float(row["Unemployment"])
        )
    )

connection.commit()

print("Combined dataset inserted into the database.")
```

```python
# Count records in the database

cursor.execute(
    "SELECT COUNT(*) FROM combined_data"
)

record_count = cursor.fetchone()[0]

print("Number of records in database:", record_count)
```

```python
# Find years where unemployment was above its average

average_unemployment = combined_data[
    "Unemployment"
].mean()

cursor.execute("""
SELECT year, education, unemployment
FROM combined_data
WHERE unemployment > ?
ORDER BY unemployment DESC
""", (float(average_unemployment),))

above_average_records = cursor.fetchall()

print(
    f"Average unemployment: "
    f"{average_unemployment:.2f}%"
)

print("\nYears with above-average unemployment:")

display(
    pd.DataFrame(
        above_average_records,
        columns=[
            "Year",
            "Education",
            "Unemployment"
        ]
    )
)
```

```python
# Select a year for the safe UPDATE demonstration

test_year = int(combined_data.iloc[0]["Year"])

cursor.execute("""
SELECT education
FROM combined_data
WHERE year = ?
""", (test_year,))

original_education = cursor.fetchone()[0]

print("Test year:", test_year)
print("Original education value:", original_education)
```

```python
# Safely update one database record

updated_value = float(original_education) + 0.01

cursor.execute("""
UPDATE combined_data
SET education = ?
WHERE year = ?
""", (updated_value, test_year))

connection.commit()

print(
    f"Education value for {test_year} "
    "updated successfully."
)
```

```python
# Verify the UPDATE

cursor.execute("""
SELECT year, education
FROM combined_data
WHERE year = ?
""", (test_year,))

updated_record = cursor.fetchone()

print("Updated record:", updated_record)
```

```python
# Restore the original value after the UPDATE demonstration

cursor.execute("""
UPDATE combined_data
SET education = ?
WHERE year = ?
""", (float(original_education), test_year))

connection.commit()

print("Original value restored successfully.")
```

```python
# Insert a temporary record for DELETE demonstration

temporary_year = 9999

cursor.execute("""
INSERT OR REPLACE INTO combined_data
(year, education, unemployment)
VALUES (?, ?, ?)
""", (temporary_year, 0.0, 0.0))

connection.commit()

print("Temporary record created.")
```
